# [MCP (Model Context Protocol)](https://modelcontextprotocol.io/docs/2026-07-28/getting-started/intro):

An open source standard/common protocol for connecting AI apps to external systems like slack, docker, email etc.

![Image](https://mintcdn.com/mcp/bEUxYpZqie0DsluH/images/mcp-simple-diagram.png?w=2500&fit=max&auto=format&n=bEUxYpZqie0DsluH&q=85&s=dc4ab238184b6c70e06e871681c921c5)

## [Architecture](https://modelcontextprotocol.io/docs/2026-07-28/learn/architecture):

- MCP Host: The AI application that coordinates and manages one or multiple MCP clients
- MCP Client: A component that maintains a connection to an MCP server and obtains context from an MCP server for the MCP host to use
- MCP Server: A program that provides context to MCP clients

For example: Visual Studio Code acts as an MCP host. When Visual Studio Code establishes a connection to an MCP server, such as the Sentry MCP server, the Visual Studio Code runtime instantiates an MCP client object that maintains the connection to the Sentry MCP server. When Visual Studio Code subsequently connects to another MCP server, such as the local filesystem server, the Visual Studio Code runtime instantiates an additional MCP client object to maintain this connection.

## MCP Server Basics:

- A program that runs on a server exposing capabilities like prompts, tools, skills over MCP protocol.
- MCP servers can execute locally or remotely. For example, when Claude Desktop launches the filesystem server, the server runs locally on the same machine because it uses the STDIO transport. This is commonly referred to as a “local” MCP server. The official Sentry MCP server runs on the Sentry platform, and uses the Streamable HTTP transport. This is commonly referred to as a “remote” MCP server.

### Layers

MCP consists of two layers:
- Data layer: Defines the JSON-RPC based protocol for client-server communication, including capability and version discovery, and core primitives, such as tools, resources, prompts and notifications.
- Transport layer: Defines the communication mechanisms and channels that enable data exchange between clients and servers, including transport-specific connection establishment, message framing, and authorization.

#### Data layer

The data layer implements a JSON-RPC 2.0 based exchange protocol that defines the message structure and semantics. This layer includes:
- Discovery: Lets clients query a server’s supported protocol versions, capabilities, and identity through the server/discover request
- Server features: Enables servers to provide core functionality including tools for AI actions, resources for context data, and prompts for interaction templates from and to the client
- Client features: Enables servers to elicit input from the user. Sampling is deprecated as of protocol version 2026-07-28.
- Utility features: Supports additional capabilities like notifications for real-time updates and progress tracking for long-running operations

#### Transport layer

The transport layer manages communication channels and authentication between clients and servers. It handles connection establishment, message framing, and secure communication between MCP participants.
MCP supports two transport mechanisms:
- Stdio transport: Uses standard input/output streams for direct process communication between local processes on the same machine, providing optimal performance with no network overhead.
- Streamable HTTP transport: Uses HTTP POST for client-to-server messages with optional Server-Sent Events for streaming capabilities. This transport enables remote server communication and supports standard HTTP authentication methods including bearer tokens, API keys, and custom headers. MCP recommends using OAuth to obtain authentication tokens.

## MCP (Model Context Protocol) Primitives

MCP defines three core server primitives: **Tools, Resources, and Prompts**. These are the building blocks an MCP server uses to expose capabilities to an AI application (MCP Host/Client).

### 1. MCP Server Primitives

| Primitive | What is it? | Sample use case | How does the MCP server expose it? | Who controls it? | Explanation |
|---|---|---|---|---|---|
| **Tools** | Executable functions that allow an AI model to perform actions or retrieve information. | A GitHub server exposes `create_issue`, `list_pull_requests`, and `merge_pull_request`. | Through `tools/list` (discovery) and `tools/call` (execution). Each tool has a name, description, input schema (JSON Schema), and optionally an output schema. | **Model-controlled** | The LLM decides when a tool is relevant and requests its execution. The MCP client handles the request and may require user approval before executing it. |
| **Resources** | Data or contextual information that an AI application can retrieve and use. | A database server exposes database schemas, table definitions, documentation, or configuration files. | Through `resources/list` (discovery) and `resources/read` (retrieval). Each resource has a URI, name, description, and optional MIME type. | **Application-controlled** | The host application decides which resources to retrieve and how to provide their contents to the model. Resources are primarily intended for reading context, not performing actions. |
| **Prompts** | Reusable prompt templates that guide the LLM through specific tasks. | A GitHub server exposes a `review_pull_request` prompt that instructs the model to examine code changes, identify bugs, and suggest improvements. | Through `prompts/list` (discovery) and `prompts/get` (retrieval). Prompts can contain arguments and return structured messages. | **User-controlled** | The user explicitly selects a prompt, often through a slash command or menu. The host then supplies the resulting messages to the LLM. |

#### Key distinction

- **Tools** = What can the AI do?
- **Resources** = What information can the AI access?
- **Prompts** = How should the AI approach a task?

### 2. Understanding the Control Hierarchy

| Primitive | Who initiates the interaction? | Who decides what happens next? | Simple analogy |
|---|---|---|---|
| Tools | LLM | LLM requests execution; client/server enforce permissions and execution. | Giving an employee the ability to perform an operation. |
| Resources | MCP Host/Application | Host decides what information to retrieve and provide. | Giving an employee access to documentation. |
| Prompts | User | User chooses which predefined workflow to invoke. | Giving an employee a predefined task template. |

### 3. Client-Side Primitives

An MCP server can also request certain capabilities from its connected MCP client. These are different from the three core server primitives.

| Primitive | What is it? | Sample use case | How is it exposed? | Who controls it? | Explanation |
|---|---|---|---|---|---|
| **Elicitation** | Allows an MCP server to request additional information from the user through the client. | A booking tool needs to know the user's preferred date before proceeding. | Server sends an `elicitation/create` request to the client. | Server initiates; user provides the input. | The server can request missing information or confirmation. The client displays the appropriate interaction to the user. |
| **Sampling** | Allows an MCP server to request an LLM completion through the client. | A server retrieves search results and asks the client's LLM to summarize them. | `sampling/createMessage` | Server initiates; client/model handles generation. | Lets servers use the client's LLM without integrating their own model provider. Deprecated in the 2026-07-28 specification. |
| **Roots** | Allows the client to tell an MCP server which filesystem locations are relevant or accessible. | A filesystem server is informed that the relevant project directory is `/workspace/my-project`. | `roots/list` | Client-controlled. | Communicates filesystem boundaries or working locations. Deprecated in the 2026-07-28 specification. |

**Important:** As of the 2026-07-28 specification, Sampling and Roots are deprecated. Existing implementations may retain compatibility, but new implementations should not depend on them. Elicitation remains supported.

### 4. Practical Example: GitHub MCP Server

Imagine building an MCP server for GitHub.

#### GitHub MCP Server

**TOOLS (Actions)**
- `create_issue()`
- `list_pull_requests()`
- `merge_pull_request()`

**RESOURCES (Context)**
- `github://repo/README.md`
- `github://repo/issues`
- `github://repo/contributors`

**PROMPTS (Templates)**
- `review_pull_request`
- `summarize_repository`
- `analyze_bug_report`

#### Example workflow

User asks:

"Look at my latest pull request, review the code and merge it if everything looks good."

Here's how the primitives could work together:

1. **Resources – Retrieve context**
   - The host can retrieve repository documentation, coding conventions, and other relevant context.

2. **Prompts – Establish the workflow**
   - The user selects `review_pull_request`, providing a predefined review structure.

3. **Tools – Perform actions**
   - The LLM calls `list_pull_requests()` and retrieves the latest PR.
   - It then calls other tools to inspect its changes.

4. **Elicitation – Request additional input (if needed)**
   - The server could ask the user for clarification or confirmation before a sensitive operation.

5. **Tools – Execute the final action**
   - After review and any required approval, the LLM requests `merge_pull_request()`.

**Important:** These primitives aren't mutually exclusive. A single MCP server can expose all three server primitives, and an application can combine them to accomplish a workflow.

### 5. Primitives vs Transport vs Protocol

An additional distinction worth remembering:

- **Primitives** define WHAT capabilities are available (Tools, Resources, Prompts).
- **Transport** defines HOW the client and server communicate (e.g., stdio or Streamable HTTP).
- **JSON-RPC** defines the message format used for communication.

## Model Context Protocol (MCP) – Complete Lifecycle

### 1. Introduction

#### High-Level Architecture

```mermaid
flowchart TD
    U[User] --> H[MCP Host / AI Application]
    H --> L[LLM]
    H --> C[MCP Client]
    C <-->|MCP Protocol| S[MCP Server]

    S --> T[Tools]
    S --> R[Resources]
    S --> P[Prompts]

    T --> E[External APIs / Services]
    R --> D[Databases / Files / Data Sources]
    P --> PT[Prompt Templates]
```


### 2. MCP Lifecycle – Three Major Phases

The MCP lifecycle can be understood in three broad phases:

1. Before Connection – Configuration and preparation.
2. During Connection – Discovery, initialization (in legacy versions), negotiation and communication.
3. After Connection – Termination and cleanup.

There are two lifecycle models depending on the protocol version:

- Traditional lifecycle: `2025-11-25` and earlier.
- Modern stateless lifecycle: `2026-07-28`.

The distinction is important because the modern protocol removes the traditional initialization handshake and protocol-level sessions.

#### Overall Lifecycle

```mermaid
flowchart TD
    A[Start] --> B[PHASE 1: Before Connection]

    B --> C[Configure MCP Server]
    C --> D[Configure MCP Client]
    D --> E[Select Transport]
    E --> F[Authorization if required]

    F --> G[PHASE 2: During Connection]

    G --> H{Protocol Version}

    H -->|2025-11-25 and earlier| I[Initialize Handshake]
    I --> J[Protocol Version Negotiation]
    J --> K[Capability Negotiation]
    K --> L[notifications/initialized]

    H -->|2026-07-28| M[Optional server/discover]
    M --> N[Per-request Metadata]
    N --> O[Stateless Request Processing]

    L --> P[Normal MCP Operations]
    O --> P

    P --> Q[Discover Tools / Resources / Prompts]
    Q --> R[Execute Operations]
    R --> S[Return Results]

    S --> T[PHASE 3: Termination]
    T --> U[Close Transport / Release Resources]
```

### 3. Phase 1 – Before Connection

Before communication begins, the MCP environment needs to be configured.

This phase is primarily about preparing the server, client, transport and authentication.

#### 3.1 Server Setup

The MCP server is implemented and configured.

It decides which capabilities it wants to expose.

For example, a GitHub MCP server might expose:

**Tools**
- `create_issue`
- `list_pull_requests`
- `merge_pull_request`

**Resources**
- `github://repository/README.md`
- `github://repository/issues`

**Prompts**
- `review_pull_request`
- `summarize_repository`

The server registers these capabilities with its MCP implementation.

#### 3.2 Client Setup

The MCP client is configured inside the MCP Host.

The client needs to know how to reach or launch the MCP server.

For example, a host configuration might look like:

```json
{
  "mcpServers": {
    "github": {
      "command": "npx",
      "args": [
        "-y",
        "github-mcp-server"
      ],
      "env": {
        "GITHUB_TOKEN": "YOUR_TOKEN"
      }
    }
  }
}
```

This is an illustrative configuration. Actual configuration formats depend on the host application.

The host reads the configuration and creates an MCP client for the server.

#### 3.3 Server Discovery / Configuration

The client needs a way to locate the server.

There is no universal MCP server discovery mechanism that every host must implement.

Common approaches:

| Approach | Explanation |
|---|---|
| Local configuration | Host is configured with the server command and arguments. |
| Remote configuration | Host is configured with the remote MCP endpoint. |
| Environment variables | Credentials and configuration are passed to the server process. |
| Service discovery | Infrastructure or deployment tooling identifies the remote server endpoint. |

#### 3.4 Transport Selection

Transport defines how the MCP client and server exchange messages.

MCP primarily uses:

| Transport | Description | Common use case |
|---|---|---|
| STDIO | Client launches the server as a subprocess and communicates using standard input and output. | Local MCP servers. |
| Streamable HTTP | Client communicates with a remote MCP server using HTTP POST requests and optional streaming responses. | Remote MCP servers. |

##### STDIO communication

```text
MCP Host
   |
   | Starts subprocess
   v
MCP Server
   |
   | JSON-RPC messages over stdin/stdout
   |
```

Important:
- The client launches the server process.
- The client writes requests to server stdin.
- The server writes responses to stdout.
- Server diagnostics should go to stderr, not stdout, because stdout carries protocol messages.

##### Streamable HTTP communication

```text
MCP Client
   |
   | HTTP POST /mcp
   v
MCP Server
   |
   | JSON-RPC request/response
   v
MCP Client
```

The server can return normal HTTP responses or streaming responses where supported.

#### 3.5 Authentication

For remote MCP servers, authorization may be required.

MCP's HTTP authorization framework uses OAuth-based mechanisms.

A simplified authorization flow:

1. Client identifies that authorization is required.
2. Client discovers the authorization server information.
3. Client obtains authorization from the user where required.
4. Client obtains an access token.
5. Client sends requests with the appropriate Authorization header.

Example:

```http
POST /mcp HTTP/1.1
Host: example.com
Authorization: Bearer ACCESS_TOKEN
Content-Type: application/json
```

Important:
- Authentication and authorization are separate from the MCP initialization lifecycle.
- STDIO servers generally use their own local credential configuration.
- Never assume that every MCP server requires OAuth.



### 4. Phase 2 – During Connection

This is where the actual MCP protocol communication happens.

There are two different models depending on the protocol version.

#### 4.1 Traditional Lifecycle – 2025-11-25 and Earlier

The traditional MCP protocol uses an initialization handshake.

The purpose is to establish compatibility between the client and server before normal operations begin.

##### Step 1: Client Sends Initialize Request

The client sends an `initialize` request.

The request communicates:

- Supported protocol version.
- Client capabilities.
- Client name.
- Client version.

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "method": "initialize",
  "params": {
    "protocolVersion": "2025-11-25",
    "capabilities": {
      "roots": {
        "listChanged": true
      },
      "sampling": {}
    },
    "clientInfo": {
      "name": "MyAIApplication",
      "version": "1.0.0"
    }
  }
}
```

Explanation:

| Field | Meaning |
|---|---|
| jsonrpc | JSON-RPC protocol version. |
| id | Request identifier used to match the response. |
| method | Operation being requested. |
| protocolVersion | Protocol version supported by the client. |
| capabilities | Optional features supported by the client. |
| clientInfo | Client name and version. |

##### Step 2: Server Sends Initialize Response

The server responds with:

- Selected protocol version.
- Server capabilities.
- Server name.
- Server version.
- Optional instructions.

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "result": {
    "protocolVersion": "2025-11-25",
    "capabilities": {
      "tools": {
        "listChanged": true
      },
      "resources": {
        "subscribe": true,
        "listChanged": true
      },
      "prompts": {
        "listChanged": true
      },
      "logging": {}
    },
    "serverInfo": {
      "name": "GitHub MCP Server",
      "version": "1.0.0"
    },
    "instructions": "Use this server to interact with GitHub repositories."
  }
}
```

##### Step 3: Capability Negotiation

Both sides communicate which optional capabilities they support.

Example:

| Client capabilities | Server capabilities |
|---|---|
| Roots | Tools |
| Sampling | Resources |
| Elicitation | Prompts |

The client can only expect server features that the server advertises.

Similarly, the server must respect the client capabilities it advertises.

Important:

Capability negotiation does not automatically grant authorization to access every external system. Permissions and access controls are separate concerns.

##### Step 4: Client Sends Initialized Notification

Once the client receives the initialization response, it sends:

```json
{
  "jsonrpc": "2.0",
  "method": "notifications/initialized"
}
```

This is a notification, not a request.

Therefore:
- It does not contain an `id`.
- The server does not send a JSON-RPC response.
- It indicates that initialization is complete.

Normal operations can now begin.

##### Traditional Handshake Sequence Diagram

```mermaid
sequenceDiagram
    autonumber
    participant H as MCP Host
    participant C as MCP Client
    participant S as MCP Server

    H->>C: Create MCP Client
    C->>S: Establish Transport

    C->>S: initialize request
    Note over C,S: Protocol version, client capabilities, clientInfo

    S-->>C: Available tools

    S-->>C: Available resources

    S-->>C: Available prompts

```



### 5. Modern Lifecycle – 2026-07-28

The 2026-07-28 specification introduces a stateless protocol core.

It removes:

- `initialize`
- `notifications/initialized`
- Protocol-level `Mcp-Session-Id` sessions

Instead, every request carries the protocol version and client capabilities in its metadata.

The new protocol introduces `server/discover` for capability discovery.

#### 5.1 Server Discovery

The client MAY call `server/discover` before making other requests.

This allows the client to discover:

- Supported protocol versions.
- Server capabilities.
- Server identity.
- Server instructions.

The server MUST implement `server/discover` under the 2026-07-28 specification.

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "method": "server/discover",
  "params": {}
}
```

The server returns discovery information.

Conceptual response:

```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "result": {
    "supportedProtocolVersions": [
      "2026-07-28",
      "2025-11-25"
    ],
    "capabilities": {
      "tools": {
        "listChanged": true
      },
      "resources": {
        "listChanged": true
      },
      "prompts": {
        "listChanged": true
      }
    },
    "serverInfo": {
      "name": "GitHub MCP Server",
      "version": "1.0.0"
    }
  }
}
```

The example illustrates the discovery concept. Exact response fields should follow the applicable specification and SDK schema.

#### 5.2 Per-Request Metadata

Instead of exchanging protocol information only once, the client includes metadata in every request.

Important metadata fields include:

| Metadata key | Purpose |
|---|---|
| `io.modelcontextprotocol/protocolVersion` | Protocol version used by the request. |
| `io.modelcontextprotocol/clientCapabilities` | Capabilities supported by the client. |
| `io.modelcontextprotocol/clientInfo` | Client identity. |
| `io.modelcontextprotocol/logLevel` | Optional logging level metadata; logging is deprecated in this specification. |

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 2,
  "method": "tools/list",
  "params": {
    "_meta": {
      "io.modelcontextprotocol/protocolVersion": "2026-07-28",
      "io.modelcontextprotocol/clientCapabilities": {},
      "io.modelcontextprotocol/clientInfo": {
        "name": "MyAIApplication",
        "version": "1.0.0"
      }
    }
  }
}
```

#### 5.3 Modern Streamable HTTP Headers

For Streamable HTTP, the new protocol also uses headers for version and routing information.

Example:

```http
POST /mcp HTTP/1.1
Host: example.com
MCP-Protocol-Version: 2026-07-28
Mcp-Method: tools/call
Mcp-Name: search
Content-Type: application/json
Authorization: Bearer ACCESS_TOKEN
```

Important headers:

| Header | Purpose |
|---|---|
| `MCP-Protocol-Version` | Identifies the protocol version. |
| `Mcp-Method` | Identifies the JSON-RPC method. |
| `Mcp-Name` | Identifies the relevant tool or operation name for routing. |
| `Authorization` | Carries authorization credentials when required. |
| `Content-Type` | Identifies the message content type. |

#### 5.4 Stateless Request Processing

The server processes each request independently.

For example:

```text
Request 1: tools/list
Request 2: resources/list
Request 3: tools/call
Request 4: prompts/list
```

Each request contains the metadata necessary for processing.

This provides important benefits:

- No protocol-level session storage.
- Easier horizontal scaling.
- Requests can be handled by different server instances.
- No sticky sessions are required at the MCP protocol layer.

Important:

Stateless MCP does not mean that the application itself must be stateless.

For example, a shopping cart workflow can create a `basket_id` in one tool call and pass that ID to subsequent tool calls.

The application can maintain state using explicit identifiers.




### 6. JSON-RPC – How MCP Messages Are Structured

MCP uses JSON-RPC 2.0 as its message format.

There are three important message types:

1. Request
2. Response
3. Notification

#### 6.1 Request

A request asks the other party to perform an operation.

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 10,
  "method": "tools/list",
  "params": {}
}
```

| Field | Meaning |
|---|---|
| jsonrpc | JSON-RPC version. |
| id | Unique request identifier. |
| method | Operation being requested. |
| params | Arguments or additional parameters. |

The receiver must respond to a valid request.

#### 6.2 Response

A response contains the result of a request.

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 10,
  "result": {
    "tools": []
  }
}
```

Important:
- The response contains the same request ID.
- A response can contain either `result` or `error`.
- It must not contain both.

#### 6.3 Notification

A notification is a message that does not require a response.

Example:

```json
{
```

### 8. Multi-Round-Trip Requests (MRTR)

The 2026-07-28 specification introduces Multi-Round-Trip Requests (MRTR).

This is an important concept because the newer protocol removes the traditional server-initiated request pattern that depended on a persistent bidirectional session.

##### Why is MRTR needed?

Imagine a user requests:

"Book a flight for me."

The MCP server needs additional information:

- Departure city
- Destination
- Travel date
- Number of passengers

Previously, the server could initiate a request to the client while processing another request.

The newer stateless protocol uses MRTR.

##### MRTR Workflow

1. Client sends an original request.
2. Server determines that additional information is required.
3. Server returns an `input_required` result.
4. The result includes `inputRequests`.
5. Client collects the requested information.
6. Client retries the original request with `inputResponses`.
7. Server continues processing.
8. Server returns a completed result.

##### MRTR Sequence Diagram

```mermaid
sequenceDiagram
    autonumber
    participant U as User
    participant H as MCP Host
    participant C as MCP Client
    participant S as MCP Server

    U->>H: Book a flight
    H->>C: Start booking tool

    C->>S: tools/call(book_flight)
    Note over S: Destination and date missing

    S-->>C: resultType = input_required
    Note over S,C: inputRequests: destination, travel date

    C->>H: Request missing information
    H->>U: Ask destination and travel date
    U-->>H: Provide requested information

    H->>C: Supply answers
    C->>S: Retry original tools/call with inputResponses

    Note over S: Continue booking operation
    S-->>C: resultType = complete
    C-->>H: Booking result
    H-->>U: Display result
```

Important:

MRTR is not simply a separate request to the server. The client retries the original request and attaches the answers using `inputResponses`.

It is the newer mechanism for multi-step interaction without requiring a persistent bidirectional protocol session.

### 9. Notifications and Events

Notifications are messages that do not require a response.

They are useful for informing the other side about changes or events.

Examples from traditional MCP:

| Notification | Purpose |
|---|---|
| `notifications/initialized` | Client informs server that initialization is complete. |
| `notifications/tools/list_changed` | Server informs client that its tool list changed. |
| `notifications/resources/list_changed` | Server informs client that its resource list changed. |
| `notifications/prompts/list_changed` | Server informs client that its prompt list changed. |
| `notifications/resources/updated` | Server informs client that a subscribed resource changed. |
| `notifications/progress` | Communicates progress for a long-running request. |
| `notifications/cancelled` | Communicates request cancellation. |
| `notifications/message` | Logging notification in older protocol versions. |

Important:

The 2026-07-28 specification changes several notification mechanisms. Logging is deprecated, and list/resource change notifications can be received through the newer subscription mechanism.

### 10. Subscriptions and Change Notifications – Modern Protocol

The 2026-07-28 protocol introduces `subscriptions/listen`.

It provides a way for clients to opt into specific categories of server-side changes.

Example:

```json
{
  "jsonrpc": "2.0",
  "id": 20,
  "method": "subscriptions/listen",
  "params": {
    "types": [
      "toolsListChanged",
      "resourcesListChanged",
      "promptsListChanged"
    ]
  }
}
```

The server acknowledges the subscription and returns subscription information.

The client can receive relevant notifications over the associated response stream.

Possible notification categories include:

- `toolsListChanged`
- `resourcesListChanged`
- `promptsListChanged`
- `resourceSubscriptions`

Important distinction:

`subscriptions/listen` is for opted-in server-to-client changes.

Request-scoped notifications such as progress and messages remain associated with the response stream of the request they relate to.

### 12. Cancellation, Progress and Long-Running Operations

MCP supports mechanisms for handling operations that take time.

#### 12.1 Progress Notifications

Suppose an MCP tool performs a large database export.

The server can report progress.

Example:

```json
{
  "jsonrpc": "2.0",
  "method": "notifications/progress",
  "params": {
    "progressToken": "export-123",
    "progress": 50,
    "total": 100
  }
}
```

This means the operation is 50% complete.

The client can display progress to the user.

#### 12.2 Cancellation

A client can request cancellation of an operation.

Example:

```json
{
  "jsonrpc": "2.0",
  "method": "notifications/cancelled",
  "params": {
    "requestId": 10,
    "reason": "User cancelled the operation"
  }
}
```

Important:

Cancellation is cooperative. Sending a cancellation notification does not guarantee that the server can immediately stop the underlying operation.

#### 12.3 Tasks Extension

For long-running operations, MCP also has a Tasks extension.

The 2026-07-28 specification moves Tasks out of the experimental core and into the official `io.modelcontextprotocol/tasks` extension.

The redesigned extension includes:

| Method | Purpose |
|---|---|
| `tasks/get` | Retrieve the current state of a task. |
| `tasks/update` | Send additional input to a task. |

Tasks allow an application to manage work that may not complete in a single immediate request.

Tasks are an extension, not a mandatory core MCP primitive.

### 13. Phase 3 – After Connection

This phase covers transport termination and resource cleanup.

There is an important distinction between the traditional and modern lifecycle models.

#### Traditional Lifecycle

In traditional stateful HTTP sessions or local STDIO processes:

1. Client stops initiating new operations.
2. Client closes the transport or terminates the subprocess.
3. Server detects the connection/process termination.
4. Server releases allocated resources.
5. Both sides clean up their resources.

For Streamable HTTP, the older stateful model could use a session termination request through HTTP DELETE.

There is no dedicated MCP JSON-RPC method named `shutdown`.

#### Modern Stateless Lifecycle

In the 2026-07-28 protocol:

- There is no protocol-level MCP session to terminate.
- Each request is independently processed.
- HTTP connections can be closed according to normal HTTP transport behavior.
- Local STDIO processes can still be terminated when the host no longer needs the server.
- The application remains free to maintain its own state using explicit handles.

Important:

Stateless protocol communication does not mean that the server process must terminate after every request.

A server can remain running and handle many independent requests.

### 16. Traditional vs Modern Lifecycle – Comparison

| Aspect | Traditional (2025-11-25) | Modern (2026-07-28) |
|---|---|---|
| Architecture | Stateful session-oriented protocol | Stateless protocol core |
| Initialization | `initialize` request | Removed |
| Initialization confirmation | `notifications/initialized` | Removed |
| Version negotiation | During initialization | Per-request metadata and optional discovery |
| Server discovery | Initialization response | `server/discover` |
| Session ID | Supported in stateful Streamable HTTP | Removed |
| Request metadata | Session-level information | Protocol version and client capabilities on every request |
| Horizontal scaling | May require shared session state or sticky routing | Requests can be handled by independent server instances |
| Server-to-client interaction | Bidirectional requests | MRTR for additional input |
| Subscription mechanism | Older resource subscriptions | `subscriptions/listen` |
| Shutdown | Close transport/session | No protocol-level session to shut down |

### Final Mental Model

Think of MCP as a communication protocol between an AI application and external capabilities.

**Before Connection**
- Configure the server.
- Configure the client.
- Select transport.
- Handle authentication where required.

**During Connection**
- Traditional: Initialize → Negotiate capabilities → Confirm initialization.
- Modern: Discover capabilities optionally → Carry metadata per request.
- Discover Tools, Resources and Prompts.
- Execute operations.
- Exchange results, notifications and additional input.

**After Connection**
- Complete outstanding work.
- Close the underlying transport when appropriate.
- Release resources.

The most important distinction:

- MCP Host manages the overall AI application.
- MCP Client handles MCP communication.
- MCP Server exposes capabilities.
- LLM decides how to use those capabilities.
- Transport moves messages between client and server.
- JSON-RPC defines the message structure.

#### 4.1 Traditional Initialization Handshake

For the traditional MCP lifecycle, the initialization handshake looks like this.

##### Step 1: Initialize

The MCP client sends:

```json
{
  "jsonrpc": "2.0",
  "id": 1,
  "method": "initialize",

    "protocolVersion": "2025-11-25",
    "capabilities": {},
```

##### Step 2: Initialization Confirmation

The client sends:

```json
{
  "jsonrpc": "2.0",
  "method": "notifications/initialized"
}
```

This is a notification, so there is no JSON-RPC response.

##### Important Version Note

### 7. What Exactly Happens Over Streamable HTTP?

Let's zoom into the network communication.

Unlike STDIO, where the client writes JSON-RPC messages to a process's stdin, Streamable HTTP uses HTTP requests.

For example, when the MCP client executes `search_repositories`, it sends an HTTP POST request.

##### HTTP Request

```http
POST https://api.githubcopilot.com/mcp/ HTTP/1.1
Authorization: Bearer YOUR_GITHUB_TOKEN
Content-Type: application/json
Accept: application/json, text/event-stream
```

##### Request Body

```json
{
  "jsonrpc": "2.0",
  "id": 10,
  "method": "tools/call",
  "params": {
    "name": "search_repositories",
    "arguments": {
      "query": "language:java spring-boot"
    }
  }
}
```

The server can respond using either:

- `application/json` – a normal JSON-RPC response.
- `text/event-stream` – an SSE stream containing JSON-RPC messages.

The client understands both.

**Important:** Streamable HTTP does not mean every request maintains a continuously open connection. Streaming is optional and depends on the server's response.

### 6. Tool 1 — create_todo

```python
@mcp.tool
def create_todo(

        str,
        Field(min_length=1, max_length=200)

) -> Todo:
    """Create a new TODO."""
```

This is where our first Python function becomes an MCP tool.

#### 6.1 What does @mcp.tool do?

The decorator registers the function with FastMCP.

It exposes the function as a tool that an MCP client can discover and invoke.

FastMCP inspects:

- Function name.
- Function docstring.
- Input parameters.
- Input type annotations.
- Return type annotation.

The resulting metadata becomes part of the tool definition.

#### 6.2 Understanding Annotated and Field

```python
title: Annotated[
    str,
    Field(min_length=1, max_length=200)
]
```

`Annotated` allows us to attach metadata to a Python type.

Here:

- The actual type is `str`.
- `Field` provides additional validation constraints.

The title must be between 1 and 200 characters.

FastMCP uses this metadata when generating the input schema.

#### 6.3 Function implementation

```python
title = title.strip()

if not title:
    raise ToolError("TODO title cannot be empty.")
```

We remove leading and trailing whitespace.

Why do we need this additional validation?

Because a title containing only spaces could satisfy a minimum string length of one before stripping.

For example:

```python
"     "
```

After stripping:

```python
""
```

We reject that value.

#### 6.4 Generate a unique ID

```python
id=uuid.uuid4().hex[:8]
```

UUID generates a unique identifier.

We use the first eight hexadecimal characters for a shorter ID.

Example:

```text
a1b2c3d4
```

For a real system, we would need to consider collision handling or use the full UUID.

#### 6.5 Generate a timestamp

```python
created_at=datetime.now(timezone.utc)
```

This creates a timezone-aware UTC timestamp.

Using UTC avoids depending on the machine's local timezone.

#### 6.6 Store the TODO

```python
with todos_lock:
    todos[todo.id] = todo
```

We store the new TODO in our dictionary.

#### 6.7 Return the result

```python
return todo
```

We return a Pydantic model rather than a dictionary or plain text.

FastMCP handles its serialization and structured output.

```json
{
    "title": "Learn FastMCP"
}
```

```json
    "title": "Learn FastMCP",

```


### 7. Tool 2 — list_todos

```python
@mcp.tool
def list_todos(
    completed: bool | None = None,
) -> TodoList:
    """List TODOs with an optional completion filter."""
```

This tool returns TODOs.

The interesting part is:

```python
completed: bool | None = None
```

This is a Python union type.

It means the argument can be:

- `True`
- `False`
- `None`

The default value is `None`.

We use this to support filtering.

| Input | Behavior |
|---|---|
| `true` | Return completed TODOs |
| `false` | Return pending TODOs |
| `null` or omitted | Return all TODOs |

#### 7.1 Filtering logic

```python
with todos_lock:
    result = [
        todo
        for todo in todos.values()
        if completed is None or todo.completed == completed
    ]
```

Let's break this down.

`todos.values()` returns all stored Todo objects.

The list comprehension iterates over each TODO.

The condition:

```python
completed is None or todo.completed == completed
```

means:

If no filter was provided, include every TODO.

Otherwise, include only TODOs whose completion status matches the supplied filter.

#### 7.2 Return a structured response

```python
return TodoList(
    todos=result,
    count=len(result),
)
```

We return both the matching TODOs and their count.

Example:

Input:

```json
{
    "completed": false
}
```

Output:

```json
{
    "todos": [
        {
            "id": "a1b2c3d4",
            "title": "Learn FastMCP",
            "completed": false,
            "created_at": "2026-11-12T10:00:00Z"
        }
    ],
    "count": 1
}
```

### 8. Resource — todos://all

```python
@mcp.resource("todos://all")
def all_todos_resource() -> str:
    """Read all TODOs as JSON."""
```

So far, we have implemented tools.

Now we expose a resource.

#### 8.1 What is an MCP resource?

An MCP resource exposes data to an MCP client.

A resource is identified by a URI.

Our resource URI is:

```text
todos://all
```

An MCP client can discover this resource and request its contents.

#### 8.2 Resource implementation

```python
with todos_lock:
    result = TodoList(
        todos=list(todos.values()),
        count=len(todos),
    )
```

We read all TODOs and construct a TodoList model.

Then:

```python
return result.model_dump_json(indent=2)
```

This serializes the Pydantic model into a JSON string.

For example:

```json
{
    "todos": [
        {
            "id": "a1b2c3d4",
            "title": "Learn FastMCP",
            "completed": false,
            "created_at": "2026-11-12T10:00:00Z"
        }
    ],
    "count": 1
}
```

#### 8.3 Why use a resource instead of a tool?

There is an important conceptual difference.

A tool represents an operation.

A resource represents data.

In our application:

| Requirement | MCP primitive |
|---|---|
| Create a TODO | Tool |
| List TODOs | Tool |
| Retrieve a TODO by ID | Tool |
| Complete a TODO | Tool |
| Delete a TODO | Tool |
| Read the complete TODO collection | Resource |

A client could also retrieve all TODOs using `list_todos`.

We expose the resource separately to demonstrate the resource primitive and how it differs from tools.

### 9. Understanding Tool Registration

Let's look at the complete registration mechanism.

```python
@mcp.tool
def create_todo(...):
    ...
```

This is not just a Python function definition.

The decorator registers the function with the FastMCP server.

Conceptually, FastMCP maintains an internal registry:

```text
MCP Server
│
├── Tool Registry
│   │
│   ├── create_todo
│   ├── list_todos
│   ├── get_todo
│   ├── complete_todo
│   └── delete_todo
│
└── Resource Registry
    │
    └── todos://all
```

When an MCP client connects and requests tool discovery, FastMCP uses this registry to provide the available tool definitions.

The client does not need to know how the functions are implemented.

It only needs their names, descriptions, and schemas.

### 10. What happens when an AI agent invokes a tool?

Let's imagine a user asks:

"Create a TODO to learn FastMCP."

The complete flow looks like this:

```mermaid
sequenceDiagram
    participant U as User
    participant A as AI Agent
    participant L as LLM
    participant C as MCP Client
    participant S as FastMCP Server
    participant D as TODO Dictionary

    U->>A: Create a TODO to learn FastMCP
    A->>L: Send user request and available tool definitions
    L-->>A: Select create_todo with title argument
    A->>C: Invoke create_todo
    C->>S: tools/call
    S->>S: Validate input
    S->>D: Store new Todo
    D-->>S: Stored Todo
    S-->>C: Structured tool result
    C-->>A: Return tool result
    A->>L: Send tool result
    L-->>A: Generate final response
    A-->>U: TODO created successfully
```

There are several important layers here.

##### Layer 1: User to AI agent

The user describes an intention in natural language.

##### Layer 2: AI agent to LLM

The agent provides the available tool definitions to the LLM.

The LLM decides whether a tool should be invoked.

##### Layer 3: LLM to agent

The LLM returns a tool selection and arguments.

For example:

```json
{
    "name": "create_todo",
    "arguments": {
        "title": "Learn FastMCP"
    }
}
```

This is an LLM tool-calling decision.

It is not itself an MCP request.

##### Layer 4: Agent to MCP client

The agent asks its MCP client or adapter to invoke the selected tool.

##### Layer 5: MCP client to MCP server

The MCP client sends a JSON-RPC request using the selected transport.

Conceptually:

```json
{
    "jsonrpc": "2.0",
    "id": 1,
    "method": "tools/call",
    "params": {
        "name": "create_todo",
        "arguments": {
            "title": "Learn FastMCP"
        }
    }
}
```

##### Layer 6: FastMCP server

FastMCP:

1. Receives the request.
2. Resolves the tool name.
3. Validates the arguments.
4. Executes the registered Python function.
5. Serializes the result.
6. Sends the tool result back.

##### Important distinction

There are two separate tool-calling mechanisms:

1. LLM tool calling: The LLM chooses which operation to execute.
2. MCP tool calling: The MCP client communicates with the MCP server to execute that operation.

They are related, but they are not the same protocol operation.

### 11. Server Entry Point

Now let's understand the final part of the server.

```python
if __name__ == "__main__":
```

This is a Python execution guard.

It ensures that the server starts only when the file is executed directly.

For example:

```bash
python todo_server.py
```

If another Python module imports `todo_server.py`, the server won't automatically start.

This is useful because we may want to import the server module without launching a server process.

#### 11.1 Argument parsing

```python
parser = argparse.ArgumentParser()

parser.add_argument(
    "--transport",
    choices=["stdio", "http"],
    default="stdio",
)

args = parser.parse_args()
```

We define a command-line argument named `--transport`.

It supports:

- `stdio`
- `http`

If the user doesn't supply a transport, STDIO is selected by default.

Examples:

```bash
python todo_server.py --transport stdio
```

```bash
python todo_server.py --transport http
```

#### 11.2 STDIO startup

```python
if args.transport == "stdio":
    mcp.run(transport="stdio")
```

This starts the MCP server using STDIO.

STDIO stands for standard input and standard output.

The MCP client launches our Python process.

Communication happens through:

- stdin
- stdout

There is no HTTP server or port involved.

The server generally remains alive for the lifetime of the client connection.

#### 11.3 HTTP startup

```python
elif args.transport == "http":
    mcp.run(
        transport="http",
        host="127.0.0.1",
        port=8000,
    )
```

This starts the MCP server using Streamable HTTP.

FastMCP starts the HTTP application and serves MCP requests.

Our endpoint becomes:

```text
http://127.0.0.1:8000/mcp
```

The HTTP server runs independently of the MCP client.

The client connects to the existing endpoint instead of launching a subprocess.

### 12. STDIO vs HTTP - Internal Architecture

The important point is that both transports use the same tool registry and business logic.

#### STDIO

```mermaid
flowchart TD
    A[Jupyter Notebook]
    B[FastMCP Client]
    C[Python Subprocess]
    D[FastMCP Server]
    E[TODO Tools]
    F[In-memory Dictionary]

    A --> B
    B -->|Launch process| C
    C --> D
    B <-->|JSON-RPC over stdin/stdout| D
    D --> E
    E --> F
```

When using STDIO:

1. Jupyter creates a FastMCP client.
2. The client launches `todo_server.py`.
3. The server registers its tools and resources.
4. The client performs the MCP initialization handshake.
5. The client discovers available tools.
6. The client invokes tools through STDIO.

#### Streamable HTTP

```mermaid
flowchart TD
    A[Jupyter Notebook]
    B[FastMCP Client]
    C[HTTP Server]
    D[FastMCP Server]
    E[TODO Tools]
    F[In-memory Dictionary]

    A --> B
    C --> D
    B <-->|HTTP requests to /mcp| C
    D --> E
    E --> F
```

When using HTTP:

1. We start the Python server independently.
2. The server listens on port 8000.
3. Jupyter creates an HTTP MCP client.
4. The client connects to `/mcp`.
5. The client performs MCP initialization.
6. The client discovers and invokes tools.

### 13. Jupyter Notebook - STDIO Client

The server is saved in `todo_server.py`.

We don't manually start the STDIO process.

The FastMCP client launches it.

```python
import sys
from pathlib import Path
from fastmcp import Client

server_path = Path("todo_server.py").resolve()

stdio_config = {
    "mcpServers": {
        "todo": {
            "command": sys.executable,
            "args": [
                str(server_path),
                "--transport",
                "stdio",
            ],
        }
    }
}

async with Client(stdio_config) as client:

    tools = await client.list_tools()

    for tool in tools:
        print(tool.name)

    result = await client.call_tool(
        "create_todo",
        {"title": "Learn MCP STDIO"}
    )

    print(result)
```

##### Understanding the configuration

```python
"command": sys.executable
```

Uses the Python interpreter running the notebook.

This is useful because it avoids accidentally launching the server using another Python environment.

```python
"args": [
    str(server_path),
    "--transport",
    "stdio",
]
```

Passes the server file and transport argument to the subprocess.

The client effectively launches:

```bash
/path/to/current/python todo_server.py --transport stdio
```

The client manages the process lifecycle.

When the client context exits, it closes the connection and cleans up the subprocess.

### 14. Jupyter Notebook — HTTP Client

For HTTP, we start the server independently.

#### Cell 1: Start HTTP server

```python
import subprocess
import sys
from pathlib import Path

server_path = Path("todo_server.py").resolve()

http_process = subprocess.Popen(
    [
        sys.executable,
        str(server_path),
        "--transport",
        "http",
    ],
    stdout=subprocess.DEVNULL,
    stderr=subprocess.PIPE,
    text=True,
)

print("HTTP server PID:", http_process.pid)
print("Endpoint: http://127.0.0.1:8000/mcp")
```

This launches a separate Python process.

The process runs Uvicorn through FastMCP's HTTP transport.

#### Cell 2: Connect to HTTP server

```python
from fastmcp import Client

async with Client(
    "http://127.0.0.1:8000/mcp"
) as client:

    tools = await client.list_tools()

    for tool in tools:
        print(tool.name)

    result = await client.call_tool(
        "create_todo",
        {"title": "Learn MCP HTTP"}
    )

    print(result)
```

Notice how different the client configuration is.

For STDIO, we provide a process configuration.

For HTTP, we provide a URL.

The tool invocation itself is identical.

#### Cell 3: Stop HTTP server

```python
http_process.terminate()
http_process.wait(timeout=5)

print("HTTP server stopped")
```

This stops the independently running HTTP process.
